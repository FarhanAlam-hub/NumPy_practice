## Introduction to Pandas

### What is Pandas?

A powerful library for data manipulation and analysis in Python. Built on top of NumPy, provides data structure like series and DataFrame.

### Why use Pandas?

- Easy data cleaning and transformation.
- Efficient handling of large datasets.
- Integration with other Python libraries like Matplotlib and Seaborn for visualization.

```python
import numpy as np
import pandas as pd

# Given NumPy array
temp_array = np.array([29.5, 31.0, 28.7, 32.2, 30.4, 27.9])

# 1. Convert NumPy array into Pandas Series
temperature = pd.Series(temp_array, name="Temperature")
print(temperature)

# 2. Report name, dtype, size, and index
print("Series Name:", temperature.name)
print("Data Type:", temperature.dtype)
print("Size:", temperature.size)
print("Index:", temperature.index)

# 3. Find the mean
print("Original Mean:", temperature.mean())

# 4. Convert Series to integer data type
temperature = temperature.astype(int)
print("\nInteger Series:")
print(temperature)

# 5. Recalculate the mean
print("Mean after conversion:", temperature.mean())
```

```python
# Using custom Indices 
salaries = [45000, 50000, 18000, 0, 20000, 30000]
deptt = ["python", "finance", "java", "sales", "hr", "marketing"]
salary_series = pd.Series(salaries, index = deptt, name= "Salary")
print(salary_series)

# indexing
print("Salary of finance deptt: ",salary_series ["finance"])
print("Salary of marketing deptt: ",salary_series ["marketing"])
print("Salary of python & java deptt: \n",salary_series [["python", "java"]])

# Slicing
print("Salary of finance - hr deptt: \n",salary_series ["python": "java"])
print("Salary from java to all deptt: \n",salary_series["java":])

```

### The iloc method

```python
# Indexing
print(salary_series.iloc[0])
print(salary_series.iloc[4])
print(salary_series.iloc[-1])
print(salary_series.iloc[1:3])
print(salary_series.iloc[[0,2,5]])

# Slicing
print(salary_series.iloc[:3])
print(salary_series.iloc[-3:])
print(salary_series.iloc[::2])
print(salary_series.iloc[::-1])
```

```python
date = np.array ([
"2026-06-01", "2026-06-02", "2026-06-03", "2026-06-04", "2026-06-05",
"2026-06-06", "2026-06-07", "2026-06-08", "2026-06-09", "2026-06-10",
"2026-06-11", "2026-06-12", "2026-06-13", "2026-06-14", "2026-06-15",
"2026-06-16", "2026-06-17", "2026-06-18", "2026-06-19", "2026-06-20"
])
stock_price = np.array([
245.5, 247.0, 246.5, 248.0, 250.0,
249.5, 251.0, 252.5, 250.5, 253.0,
254.0, 255.5, 253.5, 256.0, 257.0,
258.0, 257.5, 259.0, 260.5, 261.0
])
stock_series = pd.Series(stock_price, index=date, name = "stock price")
stock_series
```

```python
print("the avg of first 5 day is: ",stock_series.iloc[:5].mean())
print("the avg of last 5 day is: ",stock_series.iloc[-5:].mean())
print("the avg from 11 to 15 june 2026 is: ",stock_series.loc["2026-06-11" : "2026-06-15"].mean())
```
