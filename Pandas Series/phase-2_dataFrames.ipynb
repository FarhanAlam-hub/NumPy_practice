```python
# Creating DataFrames from a dictionary
data = {
    'Name' : ['John', 'Anna', 'Peter', 'Linda'],
    'Age' : [28, 34, 29, 42],
    'City' : ['New York', 'Paris', 'Berlin', 'London'],
    'Salary' : [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
df
```

```python
# Creating DataFrames from a list
data_list = [
       ['John', 28, "New York", 25000],
       ['Anna', 34, "Paris", 35000],
       ['Peter', 29, 'Berlin', 45000],
       ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
col = ['Name', 'Age', 'City', 'Salary']
df2 = pd.DataFrame(data_list,columns=col)
df2
```

### Selection and Indexing of Columns

```python
df2[["Name","Salary"]]
```

### Creating a new column

```python
df2["Designation"] = ["Doctor", "Engineer", "Doctor", "Engineer"]
df2
```

### Removing a column

```python
df2.drop(['City', "Salary"],axis = 1) # This do not remove from the memory,if we have to delete the col from memory we use inplace = True
```

## Conditional Selection

```python
df2
```

```python
# I only want to see those people whose age is above 30
df2[df2["Age"] > 30]
```
