# Question 2 -- NumPy operations

```python
a = np.array([10,20,30,40])
b = np.array([1,2,3,4])

print(f"The sum of two array is:{a + b}")
print(f"The diff of two array is:{a - b}")
print(f"The product of two array is:{a * b}")
print(f"The division of two array is:{a / b}")

print(a + 5)
print(a * 2)
```

# Question 3 NumPy Methods

```python
a = np.array([10, 20, 30, 40, 50])

# LEVEL --- 1
print(f"The original array is: {a}")
print(f"The sum of all element is: {np.sum(a)}")
print(f"The mean of all element is: {np.mean(a)}")
print(f"The maximum value of the array is: {np.max(a)}")
print(f"The minimum value of the array is: {np.min(a)}")

# LEVEL --- 2
print(f"The index of the max value: {np.argmax(a)}")
print(f"The index of the min value: {np.argmin(a)}")

# LEVEL --- 3

marks = np.array([78,65,92,55,88,71])

print(f"\nThe marks students got : {marks}")
print(f"The total marks is : {np.sum(marks)}")
print(f"The average marks is : {np.mean(marks)}")
print(f"The highest marks is : {np.max(marks)}")
print(f"The lowest marks is : {np.min(marks)}")
print(f"the index of the student who got the highest marks : {np.argmax(marks)}")
print(f"the index of the student who got the lowest marks : {np.argmin(marks)}")
print(f"the standard deviation : {np.std(marks)}")



```
