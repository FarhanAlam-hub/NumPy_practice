# question 1

```python
arr = np.arange(1,13) .reshape(3, 4)
print(arr)
print(f"First row: {arr[0]}")
print(f"Second column: {arr[: , 1]}")
print(f"Sixth Element: {arr[1 , 1]}")
print(f"First Two Row: \n {arr[:2]}")
print(f"Last 2 column: \n {arr[:,-2:]}")
print(f"Reverse the array: \n {arr[::-1]}")


```

# Question 2 -- NumPy operations

```python
a = np.array([10,20,30,40])
b = np.array([1,2,3,4])

print(f"The sum of two array is:{a + b}")
print(f"The diff of two array is:{a - b}")
print(f"The product of two array is:{a * b}")
print(f"The division of two array is:{a / b}")

print(a + 5)
print(a * 2)
```
