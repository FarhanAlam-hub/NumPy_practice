## Advance operation with Business Examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np. array([
        [1, 150000, 180000, 220000, 250000], # Paradise Biryani
        [2, 120000, 140000, 160000, 190000], # Beijing Bites
        [3, 200000, 230000, 260000, 300000], # Pizza Hub
        [4, 180000, 210000, 240000, 270000], # Burger Point
        [5, 160000, 185000, 205000, 230000]  # Chai Point
])

print("==== Zomato sales analysis ====")
print("\n Sales data shapes",sales_data.shape)
print("\n Sample data for first 3 resturant: ",sales_data[0:3])
```

```python
# total sales per year
print(np.sum(sales_data,axis=0))
year_total = np.sum(sales_data[:,1:], axis = 0)  # axis=0 — Column ke hisaab se calculation
print(year_total)

# Minimum sales per resturant
min_sales = np.min(sales_data[:,1:], axis=1)    # axis=1 — Row ke hisaab se calculation
print(f"Minimum sales of each resturant: {min_sales}")

# Maximum sales per year
max_sales = np.max(sales_data[:,1:],axis=0)
print(max_sales)

# Average sales per resturant
avg_sales = np.mean(sales_data[:,1:], axis = 1)
print(avg_sales)

# Cumalative sales per resturant
cumsum = np.cumsum(sales_data[:,1:], axis= 1)
print(cumsum)
```

```python
plt.Figure(figsize=(4, 4))
plt.plot(np.mean(cumsum, axis = 0))
plt.title("Average cumulative sales across all resturant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector Addition: ", vector1 + vector2)
print("\n Vector Multiplication: ", vector1 * vector2)
print("\n Dot Product: ",np.dot(vector1,vector2))

angle = np.arccos(np.dot(vector1,vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
