## Numpy array and basics

```python
import numpy as np
```

### Creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print(f"1D array: {arr_1d}")

arr_2d = np.array([[1,2],[3,4]])
print(f"2D array: {arr_2d}")
```

### List vs numpy array

```python
py_list = [1,2,3]
print(f"List multiplication is: {py_list * 2}")

py_arr = np.array([1,2,3])
print(f"Array multiplication is: {py_arr * 2}")
```

### Creating Array from Scratch

```python
zeros = np.zeros((3,4))
print(f"Zeros array :\n {zeros}")

ones = np.ones((3,3))
print(f"ones array :\n {ones}")

full = np.full((3,3), 7)
print(f"full array :\n {full}")

random = np.random.random((2,2))
print(f"random array :\n {random}")

sequence = np.arange(0,11,2)
print(f"sequence array :\n {sequence}")
```
